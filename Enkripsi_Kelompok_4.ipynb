```python
import time

ALFABET = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"


# ==========================================
# 1. FUNGSI ALGORITMA KRIPTO KLASIK (4 LAPIS)
# ==========================================
def enkripsi_vigenere(teks, kunci):
    hasil = ""
    panjang_kunci = len(kunci)
    for i, huruf in enumerate(teks):
        if huruf in ALFABET:
            geser = ALFABET.index(kunci[i % panjang_kunci])
            indeks = (ALFABET.index(huruf) + geser) % 26
            hasil += ALFABET[indeks]
    return hasil


def enkripsi_pbox(teks):
    # Menambahkan padding 'X' jika panjang teks bukan kelipatan 4
    sisa = len(teks) % 4
    if sisa != 0:
        teks += "X" * (4 - sisa)

    hasil = ""
    # Aturan P-Box: 1->2, 2->4, 3->1, 4->3
    # Dalam indeks Python (mulai dari 0): posisi baru diisi oleh [indeks 2, indeks 0, indeks 3, indeks 1]
    for i in range(0, len(teks), 4):
        blok = teks[i : i + 4]
        blok_baru = blok[2] + blok[0] + blok[3] + blok[1]
        hasil += blok_baru
    return hasil


def enkripsi_affine(teks, a, b):
    hasil = ""
    for huruf in teks:
        if huruf in ALFABET:
            x = ALFABET.index(huruf)
            indeks = (a * x + b) % 26
            hasil += ALFABET[indeks]
    return hasil


def enkripsi_caesar(teks, geser):
    hasil = ""
    for huruf in teks:
        if huruf in ALFABET:
            indeks = (ALFABET.index(huruf) + geser) % 26
            hasil += ALFABET[indeks]
    return hasil


# ==========================================
# 2. FUNGSI KOMPOSIT (PRODUCT CIPHER)
# ==========================================
def enkripsi_komposit(plaintext, kunci_vig, a_aff, b_aff, kunci_caesar):
    # Pastikan teks bersih (huruf kapital tanpa spasi)
    teks_bersih = "".join([c for c in plaintext.upper() if c in ALFABET])

    # Lapis 1: Vigenère
    c1 = enkripsi_vigenere(teks_bersih, kunci_vig)
    # Lapis 2: P-Box (Blok 4 huruf)
    c2 = enkripsi_pbox(c1)
    # Lapis 3: Affine
    c3 = enkripsi_affine(c2, a_aff, b_aff)
    # Lapis 4: Caesar
    cipher_final = enkripsi_caesar(c3, kunci_caesar)

    return cipher_final


# ==========================================
# 3. EKSEKUSI & PENGUKURAN RUNTIME
# ==========================================
# Uji coba menggunakan plaintext manual Anda ("KODEAMAN") -> hasil harus "BNMIOITB"
# Ganti variabel teks_asal di bawah ini dengan plaintext dari dosen ("RAHASIAKELOMPOKANDAADALAH...")
teks_asal = " "

KUNCI_VIGENERE = " "
A_AFFINE = 7
B_AFFINE = 10
KUNCI_CAESAR = 4

waktu_mulai = time.time()
ciphertext_akhir = enkripsi_komposit(
    teks_asal, KUNCI_VIGENERE, A_AFFINE, B_AFFINE, KUNCI_CAESAR
)
waktu_selesai = time.time()

runtime_enkripsi = waktu_selesai - waktu_mulai

print("=== HASIL ENKRIPSI KELOMPOK 4 ===")
print(f"Teks Asal      : {teks_asal}")
print(f"Sandi Akhir    : {ciphertext_akhir}")
print(f"Waktu Eksekusi : {runtime_enkripsi:.6f} detik")
```
